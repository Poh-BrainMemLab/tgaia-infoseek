# Overview
- Inputs: Datasets to be cleaned from excel artefacts
- Output: .pkl file without artefacts

```python
import pandas as pd
import re
from itables import show
```

# Define functions

```python
def textclean(text):
    text = text.replace("_x000D_", "")
    text = text.replace("\n", " ")
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

# Clean data

```python
tgaia_sample = pd.read_excel("selected_tgaia.xlsx")
tgaia_sample = tgaia_sample.rename(columns={'prompt': 'uncleaned_prompt', 'response': 'uncleaned_response'})
show(tgaia_sample[['uncleaned_prompt','uncleaned_response']].head(10))
tgaia_sample.info()
```

```python
tgaia_sample['prompt'] = tgaia_sample['uncleaned_prompt'].apply(textclean)
tgaia_sample['response'] = tgaia_sample['uncleaned_response'].apply(textclean)
show(tgaia_sample[['prompt','response']].head(10))
tgaia_sample = tgaia_sample.drop(columns=['uncleaned_prompt','uncleaned_response'])
```

```python
tgaia_sample.to_pickle("selected_tgaia.pkl")
```
